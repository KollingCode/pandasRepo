```python
import numpy as np
import pandas as pd

arr = np.array([24 , 76, 90, 34, 34])
### Changing index values (label)
ser = pd.Series(arr , index = ["A" , "B" , "C" ,"D", "F"])
ser['B']
```

```python
import numpy as np
import pandas as pd

cities = {"lg" : "Lagos" , "ab" : "Abuja" , "kn" : "Kano"}
### Creating A Series Object
ser = pd.Series(cities , index=["a" , "b" , "c" , "d" , "e"])
ser["a"] = "Lagos"
ser["b"] = "Ogun"
ser["c"] = "Osun"
ser["d"] = "Oyo"
ser["e"] = "Ondo"


ser
```
